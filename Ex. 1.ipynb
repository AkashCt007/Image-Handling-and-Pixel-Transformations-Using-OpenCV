1. Read the image ('Eagle_in_Flight.jpg') using OpenCV imread() as a grayscale image.

7. Crop the image to extract any specific (Eagle alone)  object from the image.

10. Read in the image ('Apollo-11-launch.jpg').

```python
apollo = cv2.imread("Apollo-11-launch.jpg")
```

11. Add the following text to the dark area at the bottom of the image (centered on the image):

```python
text = 'Apollo 11 Saturn V Launch, July 16, 1969'
font_face = cv2.FONT_HERSHEY_PLAIN
font_scale = 1.5
thickness = 2

text_size = cv2.getTextSize(text, font_face, font_scale, thickness)[0]
text_x = (apollo.shape[1] - text_size[0]) // 2
text_y = apollo.shape[0] - 20

cv2.putText(apollo, text, (text_x, text_y), font_face, font_scale,
            (255, 255, 255), thickness)
```

12. Draw a magenta rectangle that encompasses the launch tower and the rocket.

```python
# rect_color = magenta
magenta = (255, 0, 255)  # BGR color
cv2.rectangle(apollo, (100, 50), (450, 800), magenta, 3)
```

13. Display the final annotated image.

```python
plt.figure(figsize=(10, 6))
plt.imshow(cv2.cvtColor(apollo, cv2.COLOR_BGR2RGB))
plt.title("Apollo 11 Launch - Annotated")
plt.axis("on")
plt.show()
```

14. Read the image ('boy.jpg').

```python
img = cv2.imread("boy.jpg")
```

15. Adjust the brightness of the image.

```python
matrix_ones = np.ones(img.shape, dtype=np.float64)
# Use 50 as the brightness adjustment amount
matrix = (matrix_ones * 50).astype(np.uint8)
```

16. Create brighter and darker images.

```python
img_brighter = cv2.add(img, matrix)
img_darker = cv2.subtract(img, matrix)
```

17. Display the images (Original Image, Darker Image, Brighter Image).

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(img_darker, cv2.COLOR_BGR2RGB))
plt.title("Darker Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(img_brighter, cv2.COLOR_BGR2RGB))
plt.title("Brighter Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

18. Modify the image contrast.

```python
# Create matrices
matrix1 = np.ones(img.shape, dtype=np.float64)
matrix2 = np.ones(img.shape, dtype=np.float64)
# Modify image contrast
img_lower = cv2.multiply(img, matrix1, scale=0.8, dtype=cv2.CV_8U)
img_higher1 = cv2.multiply(img, matrix1, scale=1.1, dtype=cv2.CV_8U)
img_higher2 = cv2.multiply(img, matrix2, scale=1.2, dtype=cv2.CV_8U)
```

19. Display the images (Original, Lower Contrast, Higher Contrast).

```python
plt.figure(figsize=(14, 4))

images = [
    (img, "Original"),
    (img_lower, "Lower Contrast (0.8)"),
    (img_higher1, "Higher Contrast (1.1)"),
    (img_higher2, "Higher Contrast (1.2)")
]

for i, (image, title) in enumerate(images):
    plt.subplot(1, 4, i + 1)
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.axis("off")

plt.tight_layout()
plt.show()
```

20. Split the image (boy.jpg) into the B,G,R components & Display the channels.

```python
b, g, r = cv2.split(img)

plt.figure(figsize=(12, 4))
for i, (channel, title) in enumerate([
    (b, "Blue Channel"), (g, "Green Channel"), (r, "Red Channel")
]):
    plt.subplot(1, 3, i + 1)
    plt.imshow(channel, cmap="gray")
    plt.title(title)
    plt.axis("off")

plt.tight_layout()
plt.show()
```

21. _merged the R, G, B , display along with orginal image 

```python
merged_bgr = cv2.merge([b, g, r])

plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(merged_bgr, cv2.COLOR_BGR2RGB))
plt.title("Merged BGR Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

22. Split the image into the H, S, V components &  Display the channels.

```python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
h, s, v = cv2.split(hsv)

plt.figure(figsize=(12, 4))
for i, (channel, title) in enumerate([
    (h, "Hue Channel"), (s, "Saturation Channel"), (v, "Value Channel")
]):
    plt.subplot(1, 3, i + 1)
    plt.imshow(channel, cmap="gray")
    plt.title(title)
    plt.axis("off")

plt.tight_layout()
plt.show()
```

23. _merged the H, S, V, display along with orginal image 

```python
merged_hsv = cv2.merge([h, s, v])
merged_rgb = cv2.cvtColor(merged_hsv, cv2.COLOR_HSV2RGB)

plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.imshow(merged_rgb)
plt.title("Merged HSV Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
